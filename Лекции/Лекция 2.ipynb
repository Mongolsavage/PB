## 0. Чего не хватает нашим программам

Всё, что мы писали на прошлой лекции, выполняется сверху вниз, ровно один раз и всегда одинаково. Такая программа умеет только одно: посчитать по формуле.

Но настоящий расчёт почти всегда звучит иначе:

- «посчитать ток, **и если** мощность выше допустимой — предупредить»;
- «посчитать то же самое **для каждого** сопротивления из списка»;
- «увеличивать нагрузку, **пока** не сработает защита».

Слова «если», «для каждого» и «пока» — это и есть тема сегодняшней лекции. Два инструмента: ветвление и цикл.

## 1. Логический тип

Начнём с вопроса, на который можно ответить только «да» или «нет».

```julia
7 > 3
```

```julia
7 < 3
```

Julia отвечает `true` или `false`. Это не текст и не число, а отдельный тип данных:

```julia
typeof(7 > 3)
```

`Bool` — логический тип. Значений у него всего два: `true` (истина) и `false` (ложь). Пишутся они маленькими буквами и без кавычек: `"true"` в кавычках — это уже строка с текстом, к логике отношения не имеющая.

Результат сравнения можно положить в переменную, как любое другое значение:

```julia
перегрузка = 7 > 3
```

## 2. Операторы сравнения

| Оператор | Смысл |
|----------|-------|
| `==` | равно |
| `!=` | не равно |
| `<` | меньше |
| `>` | больше |
| `<=` | меньше или равно |
| `>=` | больше или равно |

```julia
U = 220

U == 220
```

```julia
U != 220
```

```julia
U >= 300
```

Обратите внимание на два знака в `==`. Один знак `=` — это присваивание, «положить значение в переменную»; два знака `==` — это вопрос «а равны ли они?».

```julia
U == 220     # вопрос: равно ли? -> true
U = 380      # действие: положить 380 в U
```

### 2.1 Двойное неравенство

В большинстве языков условие «x лежит между 0 и 10» приходится писать в два сравнения. В Julia можно писать так же, как в математике:

```julia
x = 5

0 < x < 10
```

```julia
0 < 15 < 10
```

Удобно для проверки «попадает ли значение в допустимый диапазон».

## 3. Логические операции

Одного сравнения обычно мало: условия приходится соединять. Для этого есть три операции.

| Операция | Смысл | Когда истинна |
|----------|-------|---------------|
| `&&` | И | когда истинны **оба** условия |
| `\|\|` | ИЛИ | когда истинно **хотя бы одно** |
| `!` | НЕ | переворачивает: из `true` делает `false` |

```julia
true && true
```

```julia
true && false
```

```julia
false || true
```

```julia
!true
```

На практике по краям этих операций стоят не готовые `true`/`false`, а сравнения:

```julia
U = 220
I = 3.0

U > 200 && I > 2      # напряжение высокое И ток большой
```

```julia
U > 300 || I > 2      # хотя бы одно из двух
```

```julia
!(U > 300)            # неверно, что напряжение выше 300
```

### 3.1 Приоритет

Сначала выполняется `!`, затем `&&`, затем `||`.

```julia
true || false && false
```

Здесь сначала посчиталось `false && false`, и только потом `true || false`. Ответ — `true`.

Правило простое: как только в условии больше двух частей, ставьте скобки. Они ничего не стоят, а читать код становится в разы легче.

```julia
(U > 200 || U < 100) && I > 2
```

## 4. Условный оператор if
Теперь научим программу выбирать. Самая простая форма — «если условие истинно, сделай это»:

```julia
I = 3.2

if I > 3
    println("Перегрузка по току")
end
```

Разберём запись по частям:

- `if` и следом условие — то самое выражение типа `Bool`;
- дальше идёт тело — строки, которые выполнятся, только если условие истинно;
- `end` закрывает конструкцию. Без него Julia будет ждать продолжения и код не запустится.

Тело принято сдвигать вправо на четыре пробела. Julia на отступы не смотрит, ей важен только `end`, но код без отступов нечитаем — так что сдвигаем.

### 4.1 else
«Если да — одно, иначе — другое»:

```julia
I = 2.5

if I > 3
    println("Перегрузка по току")
else
    println("Ток в норме")
end
```

Ветка `else` без условия: она срабатывает во всех случаях, когда условие `if` оказалось ложным. Сработает ровно одна из двух веток — третьего не дано.

### 4.2 elseif
Когда вариантов больше двух, промежуточные условия добавляются через `elseif`:

```julia
I = 4.5

if I > 6
    println("Аварийный режим")
elseif I > 3
    println("Перегрузка")
elseif I > 1
    println("Рабочий режим")
else
    println("Холостой ход")
end
```

Условия проверяются сверху вниз, и как только одно совпало — остальные даже не смотрятся, программа уходит на `end`. Поэтому порядок веток важен: если поставить `I > 1` первой, то ток 4.5 попадёт в неё и до «Перегрузки» дело не дойдёт.

Веток `elseif` может быть сколько угодно, `else` — не больше одной, и она всегда последняя. Слово пишется слитно: `elseif`, а не `else if`.

### 4.3 Вложенные условия

Внутри ветки может стоять ещё одно ветвление:

```julia
U = 220
I = 4.0

if U > 100
    if I > 3
        println("Высокое напряжение, большой ток")
    else
        println("Высокое напряжение, ток в норме")
    end
else
    println("Низкое напряжение")
end
```

У каждого `if` должен быть свой `end` — здесь их три, и отступы помогают не запутаться, какой к какому относится.

То же самое часто можно записать короче, через `&&`:

```julia
if U > 100 && I > 3
    println("Высокое напряжение, большой ток")
end
```

## 5. Цикл for

Второй инструмент — повторение. Если нужно проделать одно и то же двадцать раз, копировать строку двадцать раз не нужно.

```julia
for i in 1:5
    println(i)
end
```

Читается так: «для i, пробегающего значения от 1 до 5, выполни тело». Структура та же, что у `if`: заголовок, тело с отступом, `end`.

Запись `1:5` — это диапазон. Он включает обе границы: и 1, и 5.

### 5.1 Диапазоны

```julia
for i in 1:3
    println("шаг ", i)
end
```

Шаг задаётся средним числом — `начало:шаг:конец`:

```julia
for R in 10:10:50      # 10, 20, 30, 40, 50
    println(R)
end
```

Обратный отсчёт — отрицательным шагом:

```julia
for i in 10:-2:0       # 10, 8, 6, 4, 2, 0
    println(i)
end
```

Шаг может быть и дробным:

```julia
for t in 0:0.5:2
    println(t)
end
```

Вместо `in` разрешается писать знак `=`, смысл тот же: `for i = 1:5`.

Пока мы ходим только по диапазонам чисел. Позже, когда появятся массивы, тот же `for` будет так же спокойно ходить по их элементам.

### 5.2 Накопление результата

Самый частый приём: завести переменную до цикла и понемногу изменять её внутри.

```julia
сумма = 0

for i in 1:10
    сумма = сумма + i
end

println("Сумма чисел от 1 до 10: ", сумма)
```

Запись `сумма = сумма + i` встречается так часто, что для неё есть сокращение:

```julia
сумма = 0

for i in 1:10
    сумма += i          # то же самое, что сумма = сумма + i
end

println(сумма)
```

Так же работают `-=`, `*=`, `/=`.

Тот же приём, но накапливаем не сумму, а максимум.

```julia
U = 220
P_max = 0.0
R_лучшее = 0

for R in 20:20:200
    P = U^2 / R
    if P > P_max
        P_max = P
        R_лучшее = R
    end
end

println("Наибольшая мощность ", round(P_max, digits = 1), " Вт при R = ", R_лучшее, " Ом")
```

Обратите внимание: `if` стоит внутри `for`. Ветвления и циклы свободно вкладываются друг в друга — это и есть основной способ строить программы.

Ещё один типовой приём — счётчик. Сколько чисел от 1 до 100 делятся на 7 без остатка:

```julia
счётчик = 0

for i in 1:100
    if rem(i, 7) == 0
        счётчик += 1
    end
end

println("Таких чисел: ", счётчик)
```

### 5.3 Вложенные циклы

Цикл внутри цикла: внешний делает один шаг, внутренний за это время прокручивается целиком.

```julia
for i in 1:3
    for j in 1:4
        print(i * j, "\t")
    end
    println()      # перевод строки после каждой строки таблицы
end
```

Получилась таблица умножения 3 на 4. Внутренний `println()` без аргументов стоит уже вне внутреннего цикла, но внутри внешнего — поэтому он срабатывает по одному разу на строку.

## 6. Где находятся переменные цикла

Переменная-счётчик создаётся самим циклом и после `end` исчезает:

```julia
for i in 1:3
    println(i)
end

i        # ошибка: UndefVarError
```

Это не недоработка языка, а защита: счётчик — внутреннее дело цикла, и случайно испортить им переменную снаружи нельзя. Если значение нужно после цикла — сохраните его в отдельную переменную, заведённую заранее.

А вот с переменными, созданными **до** цикла, есть тонкость. В блокноте такой код работает:

```julia
сумма = 0

for i in 1:5
    сумма += i
end

println(сумма)
```

Тело цикла — это отдельная область видимости, и вообще говоря Julia не разрешает менять оттуда внешние переменные просто так. Но в интерактивной среде (блокнот, REPL) действует поблажка: если переменная с таким именем уже есть снаружи, цикл меняет именно её. Тот же код, сохранённый в файл `.jl` и запущенный как программа, выдаст предупреждение.

Нам с вами это пока ничем не грозит — мы работаем в блокноте. Запомните только рабочее правило: **накопитель создаём до цикла, внутри цикла его только изменяем**. Если написать `сумма = 0` внутри цикла, она будет обнуляться на каждом шаге, и в конце вы получите не сумму, а последнее слагаемое.

## 7. Цикл while

`for` хорош, когда заранее известно, сколько будет повторений. Но часто это неизвестно: повторять надо, пока не выполнится некоторое условие. Для этого есть `while` — «пока».

```julia
n = 1

while n <= 5
    println(n)
    n += 1
end
```

Порядок работы: проверить условие → если истинно, выполнить тело → снова проверить условие → и так далее. Как только условие стало ложным, цикл заканчивается.

Заметьте: здесь никто не увеличивает `n` автоматически. Строка `n += 1` написана руками, и она обязательна. Если её убрать, `n` навсегда останется единицей, условие всегда будет истинным и цикл не кончится никогда.

Это называется зацикливанием. Выглядит так: ячейка помечена `[*]`, кружок ядра закрашен, вывод либо не появляется, либо льётся бесконечно. Лечится прерыванием: **Esc**, затем `I`, `I` (или квадратик ■ на панели).

Полезная привычка: написав `while`, сразу же посмотрите на тело и найдите глазами строку, которая приближает условие к ложному. Не нашли — цикл вечный.

Пример, где `while` уместнее `for`. Ток растёт с нуля; надо узнать, при каком значении мощность превысит 1000 Вт. Сколько шагов на это уйдёт — заранее неизвестно:

```julia
U = 220
I = 0.0

while U * I <= 1000
    I += 0.1
end

println("Мощность превысила 1000 Вт при токе ", round(I, digits = 2), " А")
```

Итого разница простая:

| | Когда брать |
|---|---|
| `for` | число повторений известно: «по всем R от 20 до 200» |
| `while` | число повторений неизвестно: «пока не превысит предел» |

## 8. break и continue

Два слова, которые вмешиваются в ход любого цикла.

`break` — немедленно выйти из цикла, не доводя его до конца. Найдём первое число, квадрат которого больше 500:

```julia
for i in 1:100
    if i^2 > 500
        println("Первое такое число: ", i)
        break
    end
end
```

Без `break` цикл честно досчитал бы до 100 и напечатал бы все подходящие числа. С `break` он останавливается на первом же.

`continue` — пропустить остаток тела и сразу перейти к следующему шагу. Напечатаем только нечётные числа:

```julia
for i in 1:10
    if rem(i, 2) == 0
        continue
    end
    println(i)
end
```

Оба слова работают и в `for`, и в `while`. Во вложенных циклах они действуют только на тот цикл, внутри которого написаны, — внешнего не касаются.

## 9. Собираем всё вместе

Задача: для набора сопротивлений посчитать ток и мощность и пометить режимы, в которых мощность выходит за допустимую.

```julia
# Таблица режимов цепи

U = 220           # напряжение, В
P_доп = 1200      # допустимая мощность, Вт

println("R, Ом\tI, А\tP, Вт\tРежим")

for R in 20:20:200
    I = U / R
    P = U * I

    if P > P_доп
        режим = 1
    else
        режим = 2
    end

    println(R, "\t", round(I, digits = 2), "\t", round(P, digits = 1), "\t", режим)
end
```

Здесь работает всё сегодняшнее сразу: цикл гонит расчёт по всем сопротивлениям, внутри него формула из прошлой лекции, ветвление ставит оценку, а вывод собран в таблицу табуляциями.

Дополним: найдём границу — наименьшее сопротивление, при котором нагрузка ещё в норме.

```julia
U = 220
P_доп = 1200
R = 1

while U^2 / R > P_доп
    R += 1
end

println("Минимальное безопасное сопротивление: ", R, " Ом")
println("Мощность при нём: ", round(U^2 / R, digits = 1), " Вт")
```
