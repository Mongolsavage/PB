## 1. Среда: JupyterLab

### 1.1 Ядро и блокнот
Блокнот (notebook) — это файл с расширением .ipynb, ваш рабочий документ. Он состоит из ячеек.

Ядро (kernel) — это запущенная в фоне Julia, которая выполняет код из ячеек. Ядро помнит всё, что вы в нём насчитали: если в одной ячейке написать R = 100, то R будет доступна во всех остальных ячейках, пока ядро живо.

Статус ядра виден в правом верхнем углу:

○ пустой кружок — ядро свободно, ждёт;
● закрашенный — ядро занято, ячейка считается.

### 1.2 Два типа ячеек
| Тип | Что внутри | Зачем |
|-----|-----------|-------|
| **Code** | код на Julia | вычисления |
| **Markdown** | форматированный текст | заголовки, пояснения, формулы |

Переключение типа: **Esc**, затем **Y** (в код) или **M** (в текст).

### 1.3 Горячие клавиши

Запуск ячейки (работает в обоих режимах):

| Клавиши | Действие |
|---------|----------|
| **Shift + Enter** | выполнить и перейти к следующей ячейке |
| **Ctrl + Enter** | выполнить и остаться в этой же |
| **Alt + Enter** | выполнить и создать новую ячейку ниже |

В командном режиме (сначала **Esc**):

| Клавиши | Действие |
|---------|----------|
| **A** | вставить ячейку **выше** (above) |
| **B** | вставить ячейку **ниже** (below) |
| **D**, **D** | удалить ячейку (два раза подряд) |
| **Z** | отменить удаление ячейки |
| **M** | сделать ячейку текстовой (markdown) |
| **Y** | сделать ячейку кодовой |
| **I**, **I** | прервать выполнение (interrupt) |
| **0**, **0** | перезапустить ядро (restart) |
| **Ctrl + S** | сохранить блокнот |

### 1.4 Когда всё зависло или запуталось

- **Прервать вычисление** (`I`, `I` или квадратик ■ на панели) — если ячейка считает слишком долго.
- **Restart Kernel** (`0`, `0`) — ядро забывает все переменные и начинает с чистого листа.
- **Restart Kernel and Run All Cells** — перезапустить и прогнать блокнот сверху вниз.

### 1.5 Порядок выполнения

Слева от каждой выполненной ячейки стоит номер: `[1]`, `[2]`, `[3]` — это порядок, в котором ячейки **фактически** запускались. Ядро выполняет ячейки в том порядке, в котором вы нажимаете Shift+Enter, а не сверху вниз.

`[*]` означает, что ячейка выполняется прямо сейчас.

### 1.6 Ошибки

Ошибка — это нормальная часть работы.

```julia
sqrt(-1)
```

Julia выведет красный текст. Читать в нём нужно **первую строку**: там тип ошибки и её описание. Здесь — `DomainError`: у отрицательного числа нет вещественного квадратного корня.

## 2. Числа и арифметика

### 2.1 Основные операции

```julia
7 + 3      # сложение
```

```julia
7 - 3      # вычитание
```

```julia
7 * 3      # умножение
```

```julia
7 / 3      # деление
```

```julia
7 ^ 3      # возведение в степень
```

Если нужна целая часть и остаток:

```julia
div(7, 2)   # целая часть от деления: 3
```

```julia
rem(7, 2)   # остаток: 1
```

```julia
7 % 2       # то же самое, что rem
```

### 2.2 Приоритет операций

Тот же, что в математике: сначала `^`, потом `*` и `/`, потом `+` и `-`.

```julia
2 + 3 * 4      # 14, а не 20
```

```julia
(2 + 3) * 4    # 20
```

```julia
-2 ^ 2         # -4: степень старше минуса
```

```julia
(-2) ^ 2       # 4
```

### 2.3 Дробные числа не точны

```julia
0.1 + 0.2
```

Результат — `0.30000000000000004`. Это не баг Julia. Дробные числа хранятся в двоичной системе с конечной точностью, и `0.1` в ней не представима точно — ровно как `1/3` не записывается точно в десятичной.

Отсюда практическое правило: **дробные числа нельзя сравнивать на точное равенство**. Сравнения разберём в следующей лекции.

### 5.6 @show — для отладки

```julia
y = 42
@show y
@show 2 + 2
```

`@show` печатает и само выражение, и его значение. Это самый быстрый способ посмотреть, что происходит внутри вашей программы, когда она ведёт себя не так, как задумано.

## 6. Собираем всё вместе

```julia
# Расчёт цепи по закону Ома
U = 220        # напряжение, В
R = 100        # сопротивление, Ом

I = U / R      # ток, А
P = U * I      # мощность, Вт

println("Исходные данные:")
println("\tU = $U В")
println("\tR = $R Ом")
println()
println("Результат:")
println("\tI = $(round(I, digits = 3)) А")
println("\tP = $(round(P, digits = 1)) Вт")
```
